In **PyTorch**, the convolutional layer is represented in the `torch.nn` module by the class [`Conv2d`](https://pytorch.org/docs/stable/generated/torch.nn.Conv2d.html) with the following parameters:
- `in_channels`: number of input channels
- `out_channels`: number of output channels
- `kernel_size`: kernel size
- `stride`: stride
- `padding`: padding
- `padding_mode`: padding mode  (`‘zeros’`, `‘reflect’`, etc.)
- `dilation`: dilation

#### `kernel_size`

**Core size**. `int` if the core is square, and a tuple of two numbers if the core is rectangular. Specifies the size of the filter used to convolve the image.

**`kernel_size=3`**

### Image Convolution

How can convolution be useful when working with images? Convolutions detect **patterns in images**—colors and color combinations, small objects. Usually, convolution values are trainable parameters of neural networks. However, there are “ready-made” convolutions tuned to specific patterns.

For example, the Sobel operator (convolution with specific parameters) is used to detect boundaries in an image. Let's apply this operator. To do this, we will not use the `torch.nn.Conv2d` class yet, but will take the corresponding function from the `torch.nn.functional` module.

We convert the image into a tensor and rearrange the dimensions in the order `(batch_size, n_channels, height, width)`, which is accepted in **PyTorch**.

```python
img_matrix.shape
```

```python
img_tensor = torch.tensor(img_matrix, dtype=torch.float)
img_tensor = img_tensor.permute(2, 0, 1).unsqueeze(0)
img_tensor.size()
```

Let's use the Sobel operator for detecting horizontal boundaries.

```python
sobel_hor = [[-1, -2, -1], [0, 0, 0], [1, 2, 1]]

# one matrix for each image channel
kernel = [[sobel_hor, sobel_hor, sobel_hor]]
kernel = torch.tensor(kernel, dtype=torch.float)
kernel.size()
```

```python
# perform the convolution with the prepared kernel
img_conv_hor = conv2d(img_tensor, kernel)
img_conv_hor = img_conv_hor.permute(0, 2, 3, 1)
img_conv_hor.size()
```

```python
plt.figure(figsize=(1.5 * 7, 1.5 * 4))
plt.imshow(torch.abs(img_conv_hor[0, :, :, 0]))
plt.show()
```

Let's do the same for detecting vertical boundaries.

```python
sobel_ver = [[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]

# one matrix for each image channel
kernel = [[sobel_ver, sobel_ver, sobel_ver]]
kernel = torch.tensor(kernel, dtype=torch.float)

img_conv_ver = conv2d(img_tensor, kernel)

img_conv_ver = img_conv_ver.permute(0, 2, 3, 1)
img_conv_ver.size()
```

```python
plt.figure(figsize=(1.5 * 7, 1.5 * 4))
plt.imshow(torch.abs(img_conv_ver[0, :, :, 0]))
plt.show()
```

Let's combine two images

```python
img_conv = torch.sqrt(img_conv_ver**2 + img_conv_hor**2)

plt.figure(figsize=(1.5 * 7, 1.5 * 4))
plt.imshow(img_conv[0, :, :, 0])
plt.show()
```

### Comparing the performance of fully connected and convolutional networks on the MNIST dataset

First, let's solve the image classification problem using a fully connected neural network from the previous seminar.

Let's download and prepare the data we already know for classification.

```python
transform = torchvision.transforms.Compose(
    [
        torchvision.transforms.ToTensor(),
        torchvision.transforms.Normalize((0.1307,), (0.3081,)),
    ]
)

mnist_train = MNIST("./mnist/", train=True, download=True, transform=transform)
mnist_val = MNIST("./mnist/", train=False, download=True, transform=transform)


train_dataloader = DataLoader(mnist_train, batch_size=64, shuffle=True)
val_dataloader = DataLoader(mnist_val, batch_size=64, shuffle=False)
```

Подготовим цикл обучения.

```python
def train(model, optimizer, n_epochs=5, wandb_project=None):
    """The function in which the neural network is trained.

    Parameters
    ----------
    model : nn.Module
        Neural network for training.
    optimizer : torch.optim.Optimizer
        Hyperparameter optimization algorithm (e.g., SGD)
    n_epochs : int, default 5
        Number of epochs for training the neural network. An epoch is a pass through the objects
        of the dataset by the optimization algorithm with hyperparameter updates.
    wandb_project : str, default None
        Name of the project in wandb for logging the training process
        (optional).
    """
    # wandb
    if wandb_project:
        import wandb

        wandb.init(project=wandb_project)
        wandb.watch(model)

    for epoch in range(n_epochs):
        # train
        model.train()
        for x_train, y_train in tqdm(train_dataloader):
            y_pred = model(x_train)
            loss = F.cross_entropy(y_pred, y_train)
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()

        # val
        if epoch % 2 == 0:
            val_loss = []
            val_accuracy = []

            # no grad
            model.eval()
            with torch.no_grad():
                for x_val, y_val in tqdm(val_dataloader):
                    y_pred = model(x_val)
                    loss = F.cross_entropy(y_pred, y_val)
                    val_loss.append(loss.numpy())
                    val_accuracy.extend(
                        (torch.argmax(y_pred, dim=-1) == y_val).numpy().tolist()
                    )

            if wandb_project:
                wandb.log(
                    {
                        "mean val loss": np.mean(val_loss),
                        "mean val accuracy": np.mean(val_accuracy),
                    }
                )

            # metrics
            print(
                f"Epoch: {epoch}, loss: {np.mean(val_loss)}, accuracy: {np.mean(val_accuracy)}"
            )
```

We will train fully connected architecture.

```python
model = nn.Sequential(
    nn.Flatten(),  # convert a 28x28 image into a 784-dimensional vector
    nn.Linear(
        28 * 28, 128
    ),  # linear layer converting a 784-dimensional vector into a 128-dimensional vector
    nn.ReLU(),  # nonlinearity
    nn.Linear(
        128, 10
    ),  # linear layer converting a 128-dimensional vector into a 10-dimensional vector
)

optimizer = torch.optim.SGD(model.parameters(), lr=0.01, fused=True)
# create an optimizer and pass the model parameters to it
```

```python
print(
    f"Training model with {sum([x[1].numel() for x in model.named_parameters()])} parameters"
)
train(model, optimizer, wandb_project=None)
```

Изменим архитектуру, добавив в неё свёрточные слои.

```python
model = nn.Sequential(
    nn.Conv2d(
        in_channels=1, out_channels=10, kernel_size=5
    ),  # add a convolutional layer with 10 kernels
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=4),  # reduce the image by a factor of 4 on each side
    nn.Flatten(),  # convert the 6x6x10 image into a 360-dimensional vector
    nn.Linear(
        6 * 6 * 10, 128
    ),  # linear layer converting a 360-dimensional vector into a 128-dimensional vector
    nn.ReLU(),  # nonlinearity
    nn.Linear(128, 10),  # digits
)

optimizer = torch.optim.SGD(model.parameters(), lr=0.01, fused=True)
```

```python
print(
    f"Training model with {sum([x[1].numel() for x in model.named_parameters()])} parameters"
)
train(model, optimizer, wandb_project=None)
```

### Task 2

Similar to the previous example, train a neural network with the following layers:

- Convolution layer with 10 kernels of size 5
- ReLU activation function
- Reduce the image by a factor of 2 (in each dimension)
- Convolution layer with 20 kernels of size 5
- ReLU activation function
- Reduce the image by a factor of 2 (in each dimension)
- Fully connected layer with 128 neurons
- ReLU activation function
- Output layer

Write the solution in two ways:
- using the `nn.Sequential` class
- as a descendant of the `torch.nn.Module` class

```python
# 1st way
model = nn.Sequential(
    nn.Conv2d(in_channels=1, out_channels=10, kernel_size=5),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2),
    nn.Conv2d(in_channels=10, out_channels=20, kernel_size=5),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2),
    nn.Flatten(),
    nn.Linear(4 * 4 * 20, 128),
    nn.ReLU(),
    nn.Linear(128, 10),
)
```

```python
# 2nd way


class ConvNetMNIST(nn.Module):

    def __init__(self):
        """Creating a class object. Additional arguments can be used to
        change the parameters of network layers."""
        # initialize parent
        super().__init__()

        # components of the network go in the attributes
        self.conv_1 = nn.Conv2d(in_channels=1, out_channels=10, kernel_size=5)
        self.conv_2 = nn.Conv2d(in_channels=10, out_channels=20, kernel_size=5)
        self.relu = nn.ReLU()  # used 3 times
        self.pool = nn.MaxPool2d(kernel_size=2)  # used twice
        self.flatten = nn.Flatten()
        self.fc_1 = nn.Linear(4 * 4 * 20, 128)
        self.fc_2 = nn.Linear(128, 10)

    def forward(self, x: torch.tensor) -> torch.tensor:
        """Forward pass"""
        pred = self.conv_1(x)
        pred = self.relu(pred)
        pred = self.pool(pred)
        pred = self.conv_2(pred)
        pred = self.relu(pred)
        pred = self.pool(pred)
        pred = self.flatten(pred)
        pred = self.fc_1(pred)
        pred = self.relu(pred)
        pred = self.fc_2(pred)

        return pred


model = ConvNetMNIST()
```

The code written using the first method is shorter and easier to read. However, not all architectures can be represented as a sequential application of operators from `torch.nn`. Also, some operators may be repeated. Therefore, neural networks are more often written using the second method, with the help of the `nn.Module` inheritance class.

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
print(
    f"Training model with {sum([x[1].numel() for x in model.named_parameters()])} parameters"
)

train(model, optimizer)
```

```python
def train(model, optimizer, n_epochs=5, wandb_project=None):
    """The function in which the neural network is trained.

    Parameters
    ----------
    model : nn.Module
        Neural network for training.
    optimizer : torch.optim.Optimizer
        Hyperparameter optimization algorithm (e.g., SGD)
    n_epochs : int, default 5
        Number of epochs for training the neural network. An epoch is a pass through the objects
        of the dataset by the optimization algorithm with hyperparameter updates.
    wandb_project : str, default None
        Name of the project in wandb for logging the training process
        (optional).
    """

    if wandb_project:
        import wandb

        wandb.init(project=wandb_project)
        wandb.watch(model)

    for epoch in range(n_epochs):

        model.train()
        for x_train, y_train in tqdm(train_dataloader):
            y_pred = model(x_train)
            loss = F.cross_entropy(y_pred, y_train)
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()

        if epoch % 2 == 0:
            val_loss = []
            val_accuracy = []

            model.eval()
            with torch.no_grad():
                for x_val, y_val in tqdm(val_dataloader):
                    y_pred = model(x_val)
                    loss = F.cross_entropy(y_pred, y_val)
                    val_loss.append(loss.numpy())
                    val_accuracy.extend(
                        (torch.argmax(y_pred, dim=-1) == y_val).numpy().tolist()
                    )

            if wandb_project:
                wandb.log(
                    {
                        "mean val loss": np.mean(val_loss),
                        "mean val accuracy": np.mean(val_accuracy),
                    }
                )

            print(
                f"Epoch: {epoch}, loss: {np.mean(val_loss)}, accuracy: {np.mean(val_accuracy)}"
            )
```
